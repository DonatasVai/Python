```python
import random

def didesni_nei(riba) :
  skaiciuotuvas = 0
  skaiciai = ""

  for idx in range(300) :
    skaicius = random.randint(0, 300)

    skaiciai += str(skaicius) + " "

    if skaicius > riba :
      skaiciuotuvas += 1

  return skaiciuotuvas, skaiciai

rezultatas = didesni_nei(100)
print(f"Didesnių nei 100 yra: {rezultatas[0]}", rezultatas[1])
```

```python
def darbuotojai(pareigos, *vardai, **papildomi) :
  filtruoti_vardas = []

  for vardas in vardai :
    if papildomi["filtras"] not in vardas.lower() :
      filtruoti_vardas.append(vardas)

  print(f"Darbuotojai užimantys pareigas: {pareigos}, ", filtruoti_vardas)

darbuotojai("Duomenų analitikai", "Garbielė", "Antanas", "Aleksandras", "Kęstutis", filtras = "a")
```
