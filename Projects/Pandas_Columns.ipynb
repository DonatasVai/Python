```python
# Modulio importavimas
import pandas

data = pandas.read_csv("pardavimai-duomenys.csv")

data
```

```python
data["naujasStulpelis"] = "Kažkokia reikšmė"

data
```

```python
data["totalPrice"] = data["orderPrice"] + data["shippingPrice"]

data
```

```python
data.info()
```

```python
data["info"] = data["buyerCity"] + " " + data["type"]

data
```

```python
data["year"] = type(data["orderDate"])

data
```

```python
def metai(eilute) :
  return eilute["orderDate"].split("-")[0]

data["year"] = data.apply(metai, axis = 1)

data
```

```python
data["year"] = data.apply(lambda l : l["orderDate"].split("-")[0], axis = 1)

data
```

```python
def datos_ivertinimas(eilute) :
  metai = int(eilute["year"])

  if metai > 2020 :
    return "Naujas užsakymas"
  elif metai == 2020 :
    return "Covidas"
  else :
    return "Senas užsakymas"

data["datosIvertinimas"] = data.apply(datos_ivertinimas, axis = 1)

data
```

```python
def vardo_istraukimas(eilute) :
  return eilute["buyer"].split(" ")[0]

data["buyerName"] = data.apply(vardo_istraukimas, axis = 1)

data
```

```python
def lokacijos_nustatymas(eilute) :
  if eilute["buyerCity"] in ["Kaunas", "Vilnius", "Klaipėda"] :
    return "Didmiestis"
  else :
    return "Regionas"

data["lokacija"] = data.apply(lokacijos_nustatymas, axis = 1)

data
```

```python
data["lokacija"] = data.apply(lambda e : "Didmiestis" if e["buyerCity"] in ["Kaunas", "Vilnius", "Klaipėda"] else "Regionas", axis = 1)

data
```

### Užduotis 5

Sukurkite naują stulpelį, kur vertinimų balai būtų pakeisti į žodžius, pvz: 1 į "labai blogai", 2 į "blogai" ir pan.:

```python
# def vertinimai(buyerSatisfaction):
#   if buyerSatisfaction == 1:
#     return "labai blogai"
#   if buyerSatisfaction == 2:
#     return "blogai"
#   if buyerSatisfaction == 3:
#     return "vidutiniskai"
#   if buyerSatisfaction == 4:
#     return "gerai"
#   else:
#     return "puikiai"

# data["vertinimu balai"] = data['buyerSatisfaction'].apply(vertinimai)

# data





def ivertinimas(eilute) :
  if eilute["buyerSatisfaction"] == 1 :
    return "labai blogai"
  if eilute["buyerSatisfaction"] == 2 :
    return "blogai"
  if eilute["buyerSatisfaction"] == 3 :
    return "patenkinamai"
  if eilute["buyerSatisfaction"] == 4 :
    return "gerai"
  if eilute["buyerSatisfaction"] == 5 :
    return "labai gerai"

data["ivertinimas"] = data.apply(ivertinimas, axis = 1)

data[["buyerSatisfaction", "ivertinimas"]]
```

### Užduotis 6

Sukurkite naują stulpelį, kur pagal siuntimo kainą, būtų išvesta ar siuntimas yra nemokamas, ar pigus (iki 5 eurų) ar brangesnis (5 eurai ir daugiau):

```python
# def siuntimo_kaina(shippingPrice):
#   if shippingPrice == 0:
#     return "siuntimas nemokamas"
#   if shippingPrice <= 5:
#     return "pigus"
#   if shippingPrice > 5:
#     return "brangesnis"


# data["siuntimo kaina"] = data['shippingPrice'].apply(siuntimo_kaina)

# data





def siuntimas(eilute) :
  if eilute["shippingPrice"] == 0 :
    return "nemokamas"
  if eilute["shippingPrice"] < 5 :
    return "pigus"
  if eilute["shippingPrice"] >= 5 :
    return "brangesnis"

data["siuntimas"] = data.apply(siuntimas, axis = 1)

data[["shippingPrice", "siuntimas"]]
```
