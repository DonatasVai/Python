# Pandas eilutės

````python
import pandas as pd
````

````python
data = pd.read_csv("pardavimai-duomenys.csv")

data
````

### Darbas su eilutėmis

*   ```iloc``` - indexed location (nurodome eilutės(ių) numerius)
*   ```loc``` - location (nurodome sąlygą pagal ką filtruojame eilutes)



````python
data.iloc[0]
````

````python
data.iloc[[0, 1, 15, 25]]
````

````python
# Dešimt eilučių nuo 15 iki 25
data.iloc[15:25]
````

````python
# Kas antra eilutė nuo 15 iki 25
data.iloc[15:25:2]
````

````python
# Užsakymai kurie yra iš Vilniaus

data.loc[data["buyerCity"] == "Vilnius"]
````

````python
# Pirmi 20 Užsakymų kurie yra iš Vilniaus

data.loc[data["buyerCity"] == "Vilnius"].head(20)
````

````python
# Didžiosios ir mažosios raidės skiriasi:

data.loc[data["buyerCity"] == "vilnius"].head(20)
````

````python
miestas = input("Įveskite miestą: ").strip().capitalize()

data.loc[data["buyerCity"] == miestas].head(20)
````

````python
data.loc[data["buyerSatisfaction"] >= 3]
````

````python
data.loc[data["shippingPrice"] != 0]
````

*   ```&``` - and operatorius
*   ```|``` - or operatorius
*   ```~``` - not operatorius



````python
# Užsakymai kurie yra iš Kauno ir įvertinti trimis balais:

data.loc[(data["buyerCity"] == "Kaunas") & (data["buyerSatisfaction"] == 3)]
````

````python
# Užsakymai kurie yra iš Vilniaus arba iš Kauno ir kurių įvertinimas yra 1

data.loc[((data["buyerCity"] == "Vilnius") | (data["buyerCity"] == "Kaunas")) & (data["buyerSatisfaction"] == 1)]
````

````python
# Užsakymai kurie nėra iš Vilniaus arba Kauno

data.loc[~((data["buyerCity"] == "Vilnius") | (data["buyerCity"] == "Kaunas"))]
````

````python
# Užsakymai kurie nėra iš Vilniaus arba Kauno

data[["buyerCity", "buyerSatisfaction", "orderPrice"]].loc[~((data["buyerCity"] == "Vilnius") | (data["buyerCity"] == "Kaunas"))]
````

````python
# Užsakymai kurie nėra iš Vilniaus arba Kauno

data.loc[~((data["buyerCity"] == "Vilnius") | (data["buyerCity"] == "Kaunas")), ["buyerCity", "buyerSatisfaction", "orderPrice"]]
````

````python
# Duomenų rikiavimas:

data.sort_values(by = "orderPrice")
````

````python
# Rikiavimas nuo didžiausio iki mažiausio

data.sort_values(by = "orderPrice", ascending = False)
````

````python
data.sort_values(by = ["buyer", "orderPrice"], ascending = [True, False])
````

````python
data[["buyerSatisfaction", "buyer", "orderPrice", "orderStatus"]].loc[
  (data["buyerSatisfaction"] > 3) &
  (data["orderStatus"] == "pristatytas")
].sort_values(
    by = ["buyer", "orderPrice"], ascending = [True, False]
).head(20)

# SELECT
#   buyerSatisfaction,
#   buyer,
#   orderPrice,
#   orderStatus
# FROM data
# WHERE buyerSatisfaction > 3 AND orderStatus = "pristatytas"
# ORDER BY buyer ASC, orderPrice DESC
# LIMIT 20;
````

### Užduotis 9

### Užduotis 10

### Užduotis 12

### Užduotis 14

### Užduotis 22

Rasti visus užsakymus, kurie sukurti 2024 metų sausio ir vasario mėnesiais.

````python
def metai(x):
  return x.split("-")[0]
def menuo(x):
  return x.split("-")[1]


data.loc[(data['orderDate'].apply(metai) == "2024") & ((data['orderDate'].apply(menuo) == "01") | (data['orderDate'].apply(menuo) == "02"))]
````

### Užduotis 23

Rasti visus užsakymus, kurių bendra vertė (kaina + siuntimas) yra tarp 100 ir 120 eurų, bei yra pristatyti.

````python
# data.loc[((data["orderPrice"] + data["shippingPrice"]) >= 100) & ((data["orderPrice"] + data["shippingPrice"]) <= 120) & (data["orderStatus"] == "pristatytas")]



data["totalPrice"] = data["orderPrice"] + data["shippingPrice"]

# duomenys.info()

data.loc[
    (data["totalPrice"] >= 100) &
    (data["totalPrice"] <= 120) &
    (data["orderStatus"] == "pristatytas")
]
````
