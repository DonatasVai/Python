## PAPILDOMAI:

Jums duotas stringas: “54 77 2 59 17 19 108”. Paimkite jame esančius skaičius ir patikrinkite juos ar jie yra pirminiai. T.y. ar natūralusis skaičius yra didesnis nei 1 ir be liekanos dalinasi tik iš savęs ir iš vieneto.
https://en.wikipedia.org/wiki/Prime_number

````python
skaiciai = "54 77 2 59 17 19 108"

for skaicius in skaiciai.split():
    skaicius = int(skaicius)
    pirminis = True

    if skaicius <= 1:
        pirminis = False
    else:
        for daliklis in range(2, skaicius):
            if skaicius % daliklis == 0:
                pirminis = False
                break

    if pirminis:
        print(skaicius, "yra pirminis")
    else:
        print(skaicius, "nėra pirminis")
````

````python
### 7. Sugeneruokite masyvą, kurio reikšmės atsitiktinės raidės A, B, C ir D, o ilgis 200. Suskaičiuokite kiek yra kiekvienos raidės.

import random

masyvas = []

for i in range(200):
    masyvas.append(random.choice(["A", "B", "C", "D"]))

A = 0
B = 0
C = 0
D = 0

for raide in masyvas:
    if raide == "A":
        A += 1
    elif raide == "B":
        B += 1
    elif raide == "C":
        C += 1
    elif raide == "D":
        D += 1

print(masyvas)
print("A:", A)
print("B:", B)
print("C:", C)
print("D:", D)
````
