### 10. Geriausias mokinys

`mokiniai = [
    ["Jonas", 8.2],
    ["Ona", 9.5],
    ["Petras", 7.8],
    ["Ieva", 9.1]
]`

Suraskite didžiausią pažymį, o tuomet pagal jo indeksą išveskite mokinio vardą.

```python
# mokiniai = [ ["Jonas", 8.2], ["Ona", 9.5], ["Petras", 7.8], ["Ieva", 9.1] ]

# Jonas = mokiniai[0][1]
# Ona = mokiniai[1][1]
# Petras = mokiniai[2][1]
# Ieva = mokiniai[3][1]


# didziausias = print(max(Jonas,Ona, Petras, Ieva))





mokiniai = [ ["Jonas", 8.2], ["Ona", 9.5], ["Petras", 7.8], ["Ieva", 9.1] ]

didziausias = max(mokiniai[0][1], mokiniai[1][1], mokiniai[2][1], mokiniai[3][1])
vardas = mokiniai[0][0]

if mokiniai[1][1] == didziausias :
  vardas = mokiniai[1][0]

if mokiniai[2][1] == didziausias :
  vardas = mokiniai[2][0]

if mokiniai[3][1] == didziausias :
  vardas = mokiniai[3][0]

print(f"Geriausiai besikomantis mokinys: {vardas}")
```
