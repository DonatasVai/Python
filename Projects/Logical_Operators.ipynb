# Loginiai patikrinimo operatoriai

## Apie loginius operatorius

Loginiai operatoriai leidžia **vienu metu tikrinti kelias sąlygas**, o ne iš eilės kaip naudojant `elif`.

Tai suteikia daugiau lankstumo, kai norime patikrinti **sudėtingesnes logikas** viename sakinyje.

### Pagrindiniai operatoriai

| Operatorius | Reikšmė | Pavyzdys | Rezultatas |
|-------------|----------|----------|------------|
| `and`       | **IR** – abi sąlygos turi būti `True` | `5 > 2 and 10 > 3` | `True` |
| `or`        | **ARBA** – bent viena sąlyga turi būti `True` | `5 > 2 or 10 < 3` | `True` |
| `not`       | **NE** – apverčia loginę reikšmę | `not (5 > 2)` | `False` |

### Ką galime tikrinti?

- Ar **užsakymo suma yra tarp ribų** (pvz. bent 10 € ir ne daugiau kaip 20 €)?  
- Ar **užsakymas iš Kauno ir jo vertė virš 100 €**?  
- Ar **užsakymas yra atšauktas ir pirkėjo įvertinimas lygus 1 balui**?  
- Ar **siuntimo kaina yra 0 € arba 2.99 €**?  
- Ar **užsakymas pristatytas į Vilnių arba Kauną, ir bendra vertė 100 € ar daugiau**?
