### Unikalūs vartotojų vardai

* Raskite visus unikalius vartotojus.
* Raskite vartotojus, esančius visuose trijuose sąrašuose.
* Raskite vartotojus, kurie yra group_a, bet nėra kituose sąrašuose.
* Raskite vartotojus, kurie yra group_b arba group_c, bet nėra group_a.
* Rezultatą paverskite surikiuotu sąrašu.

````python
# group_a = ["Jonas", "Ona", "Petras", "Ieva", "Tomas"]
# group_b = ["Ona", "Tomas", "Laura", "Mantas"]
# group_c = ["Jonas", "Laura", "Rokas", "Ieva"]

# group_a = set(["Jonas", "Ona", "Petras", "Ieva", "Tomas"])
# group_b = set(["Ona", "Tomas", "Laura", "Mantas"])
# group_c = set(["Jonas", "Laura", "Rokas", "Ieva"])

# # Raskite visus unikalius vartotojus.
# # print(group_a.union(group_b,group_c))


# # Raskite vartotojus, esančius visuose trijuose sąrašuose.
# # print(group_a.intersection(group_b,group_c))


# # Raskite vartotojus, kurie yra group_a, bet nėra kituose sąrašuose.
# # print(group_a.difference(group_b,group_c))


# # Raskite vartotojus, kurie yra group_b arba group_c, bet nėra group_a.
# print(group_b.difference,group_c.difference(group_a))


# # bendras = group_a + group_b + group_c

# # print(list(bendras))






group_a = ["Jonas", "Ona", "Petras", "Ieva", "Tomas"]
group_b = ["Ona", "Tomas", "Laura", "Mantas"]
group_c = ["Jonas", "Laura", "Rokas", "Ieva"]

bendras = group_a + group_b + group_c

print(bendras)

# Raskite visus unikalius vartotojus.
print(list(set(bendras)))

# Raskite vartotojus, esančius visuose trijuose sąrašuose.
group_a_set = set(group_a)

esantys_abiejose = group_a_set.intersection(group_b)

print(esantys_abiejose)

esantys_trijuose = esantys_abiejose.intersection(group_c)

print(esantys_trijuose.symmetric_difference(group_b + group_c))

# Raskite vartotojus, kurie yra group_a, bet nėra kituose sąrašuose.

group_a_set.difference(group_b, group_c)

sujungtas = set(group_b + group_c)

rezultatas = sujungtas.difference(group_a)

print(rezultatas)

rezultatas = list(rezultatas)

rezultatas.sort()

print(rezultatas)
````
