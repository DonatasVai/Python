## Matematiniai operatoriai

Operatorius | Pavadinimas | Pavyzdys | Paaiškinkimas
------------|-------------|----------|--------------
`+` | Sudėtis | 5 + 3 | Sudeda 5 su 3
`-` | Atimtis | 5 - 3 | Iš penkių atima 3
`*` | Daugyba | 5 * 3 | Penkis daugina iš trijų
`/` | Dalyba | 5 / 3 | Penkis dalina iš trijų
`**`| Kėlimas laipsniu | 5 ** 3 | 5 Pakelia 3 laipsniu
`%` | Liekana (modulis) | 5 % 2 | Grąžina liekaną 1
`//`| Dalyba iki sveiko skaičiaus | 5 // 3 | Dalyba iki sveiko skaičiaus (ats. 1)
